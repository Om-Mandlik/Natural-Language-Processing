**Word Embeddings (Word2Vec)**

```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["I", "love", "NLP"],
    ["I", "love", "Python"],
    ["Python", "is", "powerful"],
    ["NLP", "is", "interesting"]
]
```

```python
model = Word2Vec(sentences,window=2,vector_size=50,min_count=1)
print(model.wv["Python"])
print(model.wv.most_similar("Python"))
```

**Fast Text**

```python
sentences1 = [
    ["cat", "drinks", "milk"],
    ["dog", "drinks", "milk"],
    ["cats", "drink", "milk"],
    ["lion", "hunts", "deer"]
]
```

```python

from gensim.models import FastText
model = FastText(sentences1,window=2,vector_size=50,min_count=1)
print(model.wv["cat"])
print(model.wv.most_similar("cat"))
```
