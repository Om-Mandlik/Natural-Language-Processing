```python
import torch
```

```python
Q = torch.tensor([[2., 1.]])
K = torch.tensor([[2., 1.],
                  [1., 2.]])
V = torch.tensor([[5., 5.],
                  [1., 9.]])
```

```python
scores = torch.matmul(Q, K.T)
weights = torch.softmax(scores, dim=1)
output = torch.matmul(weights, V)
print(scores)
print(weights)
print(output)
```
